```python
# Install and setup for Mermaid rendering
from IPython.display import HTML, display
import base64

def render_mermaid_html(diagram_code, title="Diagram"):
    """
    Render Mermaid diagram in Jupyter notebook using CDN
    """
    html = f'''
    <div style="margin: 20px 0; padding: 15px; border: 1px solid #ddd; border-radius: 8px;">
        <h3 style="color: #333; margin-bottom: 15px;">{title}</h3>
        <div id="mermaid-{hash(title)}" style="text-align: center;">
            <div class="mermaid">
                {diagram_code}
            </div>
        </div>
    </div>
    '''
    
    # Include Mermaid.js library
    script = '''
    <script src="https://cdn.jsdelivr.net/npm/mermaid@10.6.1/dist/mermaid.min.js"></script>
    <script>
        mermaid.initialize({
            startOnLoad: true,
            theme: 'default',
            themeVariables: {
                fontFamily: 'Arial, sans-serif',
                fontSize: '14px'
            }
        });
    </script>
    '''
    
    display(HTML(script + html))

print("✅ Mermaid rendering functions loaded!")
```
